```python
import numpy as np
import pandas as pd
import ast
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
movies = pd.read_csv(r"C:\Users\DELL\OneDrive\Desktop\Personalized_Recommendation_System\data\tmdb_5000_movies.csv")
credits = pd.read_csv(r"C:\Users\DELL\OneDrive\Desktop\Personalized_Recommendation_System\data\tmdb_5000_credits.csv")

movies.head(), credits.head()


```

```python
credits.rename(columns={'movie_id': 'id'}, inplace=True)

# merge on id
df = movies.merge(credits, on="id")
df.head()
```

```python
df = df[['id', 'original_title', 'overview', 'genres', 'keywords', 'cast', 'crew']]
df.head()
```

```python
import ast

def safe_convert(obj):
    # If it's already a list, return it
    if isinstance(obj, list):
        return obj
    
    # Otherwise parse JSON string → Python list of dicts
    try:
        data = ast.literal_eval(obj)
        return [i['name'] for i in data]
    except:
        return []

df.head()

```

```python
df['genres'] = df['genres'].apply(safe_convert)
df['keywords'] = df['keywords'].apply(safe_convert)
df.head()

```

```python
def safe_cast(obj):
    if isinstance(obj, list):
        return obj[:3]
    try:
        data = ast.literal_eval(obj)
        return [i['name'] for i in data][:3]
    except:
        return []
        
df['cast'] = df['cast'].apply(safe_cast)
df.head()
```

```python
def get_director(obj):
    if isinstance(obj, list):  # already parsed list?
        for i in obj:
            if i.get('job') == 'Director':
                return i.get('name', '')
        return ''
    
    # otherwise parse string to list
    try:
        data = ast.literal_eval(obj)
        for i in data:
            if i['job'] == 'Director':
                return i['name']
        return ''
    except:
        return ''
        
df['director'] = df['crew'].apply(get_director)
df.head()
```

```python
# Convert list columns into space-separated strings
df['genres'] = df['genres'].apply(lambda x: " ".join(x))
df['keywords'] = df['keywords'].apply(lambda x: " ".join(x))
df['cast'] = df['cast'].apply(lambda x: " ".join(x))
df['director'] = df['director'].apply(lambda x: x.replace(" ", ""))

# Combine everything into tags
df['tags'] = df['overview'] + " " + df['genres'] + " " + df['keywords'] + " " + df['cast'] + " " + df['director']

df[['original_title', 'tags']].head()
```

```python
df['tags'] = df['tags'].fillna("")
```

```python
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()

def stem_text(text):
    y = []
    for word in text.split():
        y.append(ps.stem(word))
    return " ".join(y)

# Clean tags
df['tags'] = df['tags'].apply(lambda x: str(x).lower())
df['tags'] = df['tags'].apply(stem_text)

df[['original_title', 'tags']].head()
```

```python
from sklearn.feature_extraction.text import CountVectorizer

cv = CountVectorizer(max_features=5000, stop_words='english')

vectors = cv.fit_transform(df['tags']).toarray()

vectors.shape
```
